#### Setup & imports

```python
import torch
from sentence_transformers import SentenceTransformer

MODEL_NAME = "all-MiniLM-L6-v2"

device = "cuda" if torch.cuda.is_available() else "cpu"
device
```

#### load model

```python
model = SentenceTransformer(MODEL_NAME, device=device)
```

#### Embed clinical texts

```python
texts = [
    "Patient reports chest pain radiating to left arm.",
    "No abnormalities detected in ECG.",
    "Blood tests indicate elevated CRP suggesting infection.",
    "MRI scan reveals no acute intracranial abnormalities."
]

embeddings = model.encode(texts, convert_to_tensor=True)
embeddings.shape
```

#### cosine similarity between sentences

```python
from sentence_transformers import util

sim_matrix = util.cos_sim(embeddings, embeddings)
sim_matrix
```

#### nearest neighbour example

```python
query = "What suggests infection?"
q_emb = model.encode(query, convert_to_tensor=True)

scores = util.cos_sim(q_emb, embeddings)[0]
best_idx = torch.argmax(scores).item()

print("Query:", query)
print("Best match:", texts[best_idx])
print("Score:", scores[best_idx].item())
```
